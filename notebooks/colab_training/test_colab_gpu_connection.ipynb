# Antigravity ↔ Google Colab Pro GPU Connection Verification

This notebook verifies the live connection between Antigravity and your Google Colab T4 GPU runtime.
- **Execution Safety:** Pure inspection only. Zero model retraining, zero checkpoint modifications, zero disk writes.

```python
# 1. Check Hardware & Driver via nvidia-smi
!nvidia-smi
```

```python
# 2. Check PyTorch CUDA Environment & Memory
import torch

print("=== PYTORCH CUDA VERIFICATION ===")
print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available:  {torch.cuda.is_available()}")

if torch.cuda.is_available():
    gpu_count = torch.cuda.device_count()
    gpu_name = torch.cuda.get_device_name(0)
    total_mem_gb = torch.cuda.get_device_properties(0).total_memory / (1024**3)
    allocated_mem_gb = torch.cuda.memory_allocated(0) / (1024**3)
    reserved_mem_gb = torch.cuda.memory_reserved(0) / (1024**3)
    print(f"GPU Devices:     {gpu_count}")
    print(f"GPU Model:       {gpu_name}")
    print(f"Total VRAM:      {total_mem_gb:.2f} GB")
    print(f"Allocated VRAM:  {allocated_mem_gb:.2f} GB")
    print(f"Reserved VRAM:   {reserved_mem_gb:.2f} GB")
    print("STATUS: ANTIGRAVITY IDE SUCCESSFULLY CONNECTED TO COLAB GPU!")
else:
    print("STATUS: CPU RUNTIME DETECTED (GPU not active).")
```
